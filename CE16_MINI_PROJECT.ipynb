```python
#Model Training & Export Script (tomato_model.h5)Run the following Python script in Google Colab or a local JupyterNotebook using the PlantVillage tomato dataset to generate thetomato_model.h5 file:
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models
```

```python
# 1. Image preprocessing & data generator
datagen = ImageDataGenerator(rescale=1./255,
validation_split=0.2)
```

```python
# Load images from directory (containing Tomato_healthy &Tomato_Bacterial_spot)
train_data = datagen.flow_from_directory(
r"C:\Users\ACHI EMMANUEL\Desktop\GET_324_PROJECT\dataset\village_plant",
target_size=(224, 224),
batch_size=32,
class_mode='binary', 
subset='training'
)
```

```python
# 2. Transfer Learning model setup using MobileNetV2
base_model = MobileNetV2(weights='imagenet',
include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False
model = models.Sequential([
base_model,
layers.GlobalAveragePooling2D(),
layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy',
metrics=['accuracy'])
```

```python
# 3. Model training
model.fit(train_data, epochs=3)
```

```python
# 4. Save model weights (using native Keras format)
model.save('tomato_model.keras')
```
